```python

import numpy as np
from numpy.polynomial.polynomial import Polynomial

# Data dari tabel
periods = [2082, 2083, 2084, 2085, 2086]
values = [9198, 6022, 2521, 4031, 6219]

# Menggunakan NumPy untuk membuat polinomial
coefficients = np.polyfit(periods, values, 2)  # Regresi polinomial derajat 2
polynomial = Polynomial(coefficients[::-1])   # Konversi ke format Polynomial

# Prediksi untuk periode 2087
predicted_value = polynomial(2087)
print(f"Prediksi nilai untuk periode 2087: {predicted_value:.2f}")
```
